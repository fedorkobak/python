# Jinja

This page considers the features of the jinja API: the python structures that the package provides.

````python
import jinja2
from IPython.display import HTML
````

## Context

The context is the mapping of the variables available Jinja variables during rendering. The context is specified by the arguments of the `Template.render` method.

---

The following cell renders the temlate with context:

```python
{"value": 10}
```

````python
context = {"value": 10}
template = jinja2.Template("this is {{ value }}")
template.render(context)
````

Alternatively, the `render` method supports kwargs, each specifies a variable in the context.

````python
template.render(value="'some data'")
````

## Environment

The `jinja2.Environment` is the object that determines the behaviour of the jinja, with it you can specify varibales of the template renderer.

The `fronm_string` method of the environment allows to construct a `Template` object, which is usefull for experimenting with configuration parameters.

Check the [`jinja2.Environment`](https://jinja.palletsprojects.com/en/stable/api/#jinja2.Environment) API description section.

---

The following cell creates an environment with the 'finalize' parameter specified, which allows you to define how expressions are rendered.

In this example it just adds `"_hello"` string to each:

````python
def finalize(value):
    return value + "_hello"


env = jinja2.Environment(finalize=finalize)
print(env.from_string("{{ 'render this' }} \nand {{ 'this' }}").render())
````

## Generate HTML

You can easily generate some html using jinja templates. For example, in the following cell I have displayed python dict as html using jinja templates.

````python
data = {'apple': 3, 'banana': 5, 'orange': 2}

template = jinja2.Template('''
    <table>
        <thead>
            <tr>
                <th>Fruit</th>
                <th>Quantity</th>
            </tr>
        </thead>
        <tbody>
            {% for key, value in data.items() %}
                <tr>
                    <td>{{ key }}</td>
                    <td>{{ value }}</td>
                </tr>
            {% endfor %}
        </tbody>
    </table>
''')

html = template.render(data=data)
HTML(html)
````
